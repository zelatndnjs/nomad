```python
from dotenv import load_dotenv
import openai

load_dotenv()

client = openai.OpenAI()

messages = []
```

```python
def call_ai():
    response = client.chat.completions.create(
        model="gpt-5-nano",
        messages=messages
    )
    message = response.choices[0].message.content
    messages.append({
        "role" : "assistant",
        "content" : message
    })
    print(f"AI : {message}")
```

```python
while True:
    message = input("Send a message to the")
    if message == "quit" or message == "q":
        break
    else:
        messages.append({
            "role" : "user",
            "content" : message
        })
        print(f"User : {message}")
        call_ai()
```
